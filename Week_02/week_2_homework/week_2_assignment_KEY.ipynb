You are given a list <code>states</code> containing two randomly selected US states (excludes US territories and excludes Washington DC). Each state is denoted by its standard two-letter abbreviation as a string. 

You are also given a period beginning on <code>start</code> and ending on <code>end</code>. Dates are provided as a pandas datetime object. Use the <code>start_date</code> column of <code>df</code> to determine if the data is covered within the given period (inclusive).

For the period covered by <code>start</code> and <code>end</code>, return which of the two states has a greater number of new cases over the entire period covered by <code>start</code> and <code>end</code>.

<u>The evaluator will import and pass the DataFrame to your function, so your function should <strong>not</strong> have <code>pd.read_csv()</code> within its code block.</u>

````python
####################################################
## THIS CODE CELL WILL BE JUDGED BY THE EVALUATOR ##
####################################################

def question_1(df, states, start, end):
    ### DO NOT MODIFY ANYTHING ABOVE THIS LINE ###
    ### YOUR ANSWER MUST BE BELOW THIS LINE WITHIN THIS CODE CELL ####
    df.start_date = pd.to_datetime(df.start_date)
    if df.loc[
        (df.state == states[0]) & 
        (df.start_date >= start) & 
        (df.start_date <= end), 'new_cases'].sum() > df.loc[(df.state == states[1]) & 
        (df.start_date >= start) & 
        (df.start_date <= end), 'new_cases'].sum():
        return states[0]
    else:
        return states[1]
````

The data table contains absorbance measurements for an MTT assay against different gene KOs and an unedited control. The absorbance measurments given in the data table are already corrected for background absorbance. The data is structured such that each biological replicate is a row and each sample condition (control and gene KOs) are columns. Column headers are structured such that the control label is <code>'Control'</code> and the gene KO labels are <code>'GENE'</code> where GENE is replaced by the gene symbol, indicating which gene was KO'ed.

You are also given a string <code>gene</code> corresponding to a gene symbol for the gene that your PI has asked you to analyze.

Using an appropriate statistical test (do not assume equal variance), return the p-value resulting from the comparison of the relative absorbances for the given gene and the control. Use the mean absorbance of the control samples to determine relative absorbance.

You do not need to generate the gene. The evaluator will pass random gene symbols (one at a time) to your function to judge your answer and the generated output.

````python
####################################################
## THIS CODE CELL WILL BE JUDGED BY THE EVALUATOR ##
####################################################

def question_3(df, gene):
    ### DO NOT MODIFY ANYTHING ABOVE THIS LINE ###
    ### YOUR ANSWER MUST BE BELOW THIS LINE WITHIN THIS CODE CELL ####

    return stats.ttest_ind(
        df.Control/df.Control.mean(),
        df[gene]/df.Control.mean(),
        equal_var=False,
    ).pvalue
````

<h2>Question 4</h2>

1pt

Like with week 1's homework question 5, you're given a data table in the following directory path:

<code>/home/jovyan/_shared/bcourses-1557888-readonly/week-1-homework-data/</code>

The data table contains absorbance measurements for an MTT assay against different gene KOs and an unedited control. The absorbance measurments given in the data table are already corrected for background absorbance. The data is structured such that each biological replicate is a row and each sample condition (control and gene KOs) are columns. Column headers are structured such that the control label is <code>'Control'</code> and the gene KO labels are <code>'GENE'</code> where GENE is replaced by the gene symbol, indicating which gene was KO'ed.

You are also given a string <code>gene</code> corresponding to a gene symbol for the gene that your PI has asked you to analyze.
 
Return an annotated swarm plot that visualizes the relative absorbances for the control and selected gene that your PI asked you to analyze. Use the mean absorbance of the control samples to determine relative absorbance.

Your returned plot needs to contain the following:

<ul>
    <li>Appropriate X-axis labels</li>
    <li>Appropriate Y-axis label</li>
    <li>Y-axis with 0 set to the origin</li>
    <li>Individual biological replicates represented as dots</li>
    <li>Overlaid descriptive statistics corresponding to the mean and standard error of the mean</li>
    <li>Annotations indicating degree of significance</li>
</ul>

You do not need to generate the gene. The evaluator will pass random gene symbols (one at a time) to your function to judge your answer and the generated output. You do not need to export the plot as a separate file.

````python
####################################################
## THIS CODE CELL WILL BE JUDGED BY THE EVALUATOR ##
####################################################

def question_4(df, gene):
    ### DO NOT MODIFY ANYTHING ABOVE THIS LINE ###
    ### YOUR ANSWER MUST BE BELOW THIS LINE WITHIN THIS CODE CELL ####


    #Answers will vary
    dataset = df[['Control', gene]]/df.Control.mean()

    conclusion = stats.ttest_ind(
        dataset.Control,
        dataset[gene],
        equal_var=False,
        nan_policy='omit',
    )
    
    fig, ax = plt.subplots()

    dot_colors = ['#777777', '#E64B35']
    
    sns.swarmplot(
        data=dataset,
        s=12,
        palette=dot_colors,
        zorder=0,
    )
    
    sns.barplot(
        data=dataset,
        estimator=np.mean,
        alpha=0,
        errorbar='se',
        capsize=0.3,
        err_kws={
            'linewidth': 1,
            'color': 'black',
        },
    )
    
    sns.boxplot(
        data=dataset,
        showmeans=True,
        meanline=True,
        width=0.5,
        meanprops={
            'color': 'black',
            'linestyle': '-',
            'linewidth': 1,
        },
        medianprops={'visible': False},
        whiskerprops={'visible': False},
        showfliers=False,
        showbox=False,
        showcaps=False,
    )
    
    x1, x2 = 0, 1
    y = dataset.max().max()
    adjustment = y*0.1
    y_max = y + adjustment
    h = adjustment*0.5
    
    plt.plot(
        [x1, x1, x2, x2],
        [y_max, y_max + h, y_max + h, y_max],
        c='k',
        lw=1,
    )
    
    if conclusion.pvalue < 0.0005:
        pvalue_text = '***'
    elif conclusion.pvalue < 0.005:
        pvalue_text = '**'
    elif conclusion.pvalue < 0.05:
        pvalue_text = '*'
    else:
        pvalue_text = 'ns'
    
    plt.text(
        (x1 + x2)*0.5,        
        y_max + (h*1.5),      
        pvalue_text,          
        ha='center',          
        va='bottom',          
        color='k',            
        size=12,              
    )
    
    ax.set_ylabel(
        'Relative corrected mean absorbance',
        size=14,
    )
    
    ax.set_xticks(
        [0,1],
        ['Control', gene],
    )
    
    plt.yticks(
        size=12,
    )
    
    plt.xticks(
        rotation=35,
        ha='right',
        size=14,
    )
    
    fig.set_dpi(300)
    
    sns.despine()
    
    plt.show()
````
