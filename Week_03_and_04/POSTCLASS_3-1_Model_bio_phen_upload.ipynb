`````python
#Initialize our arrays
#One will contain our values that we pull from
input_array = np.arange(0, 20, 4)

#And the other will hold our outputs from our for-loop
output_array = np.zeros(5)
`````

`````python
#We can run our for-loop
for i in range(0, 5, 1):
    #Take a look at what position we're pulling from and placing our output
    print(i)

    #We pull from our input_array, perform a math operation, then assign the value to our output_array
    output_array[i] = input_array[i] + 2

    #Can see how our output_array updates with each iteration of our for-loop
    print(input_array)
    print(output_array)
`````

How does your array holding each of the outputs look like?

`````python
print(input_array)
print(output_array)
`````



`````python
#We can initialize our array to hold outputs
fib_series = np.zeros(25)

#And then since the first position is already 0
#We only need to update the 2nd position to be 1
fib_series[1] = 1

#Then we run our for-loop to generate our fibonacci sequence
for i in range(0, len(fib_series) - 2, 1):
    fib_series[i + 2] = fib_series[i] + fib_series[i + 1]

fib_series.astype(int)
`````

<h1 style="font-size: 40px; margin-bottom: 0px;">Exercise #6 - Solve the equation for a line</h1>

<hr style="margin-left: 0px; border: 0.25px solid; border-color: #000000; width: 800px;"></hr>

Because for-loops repeatedly perform a set operation through a defined index, we can use it to solve mathematical equations (over a defined interval). 

For this exercise, recall the mathematical equation for a line:

````
y = mx + b
````

In a simplified view, we can provide Python with a bunch of values contained within a compound data type, such as a list or array or range. These values can then act as x values. We can have our for-loop iterate through all these values, and then with given constants for slope and the y-intercept, we can then solve for y at each x value. With an output array that holds our outputs from our for-loop, we then have the y values corresponding to each x value. 

See if you can translate the equation for a line into something that Python would understand. Then initialize a for-loop to solve an equation from x=0 to x=30 (inclusive) given the following:

<ul>
    <li>X values are integers</li>
    <li>The slope of the line is 0.75</li>
    <li>The y-intercept is 5</li>
</ul>

`````python
#Initialize our output array with just zeros
y = np.zeros(31)

#Set up our for-loop
for x in range(0, len(y), 1):
    #And then solve for our equation by calculating the values at each position
    y[x] = 0.75 * x + 5

print(y)
`````

Let's plot our results real quick to see how they look.

`````python
plt.plot(
    y,
)

plt.xlim(0)
plt.ylim(0)
`````

Recall the general equation for a polynomial:

```
y = Ax**n + Bx**n-1 + ... + Z
```

Feel free to make it as simple or as complicated of a polynomial as you want to see how Python handles it. Then plot the output.

`````python
#Set up our input and our output
x = np.arange(0, 300, 1)
y = np.zeros(len(x))

#Solve our equation with a for-loop
for i in x:
    y[i] = 2*x[i]**3 - 4*x[i]**2 + x[i] - 5


#Plot the result
fig, ax = plt.subplots()

plt.plot(x, y)

plt.ticklabel_format(axis='y', style='plain')

plt.xlim(0)
plt.ylim(0)
`````

<h1 style="font-size: 40px; margin-bottom: 0px;">Exercise #9 - Simulating a dynamic process over time</h1>

<hr style="margin-left: 0px; border: 0.25px solid; border-color: #000000; width: 975px;"></hr>

Now that we have a better understanding of how to use for-loops to solve equations, we can begin to apply our knowledge to simulating dynamic processes. For this exercise, you'll expand on what you've learned in lecture, where we translated a basic descsriptive statement of bacterial growth into a mathematical expression.

```
The change in the number of bacteria over some period of time is dependent on the number of bacteria that actually divide.
```

So translating this into a mathematical expression:
<ul>
    <li>The change in the number of bacteria &rarr; <strong>dN</strong></li>
    <li>over &rarr; <strong>/</strong></li>
    <li>some period of time &rarr; <strong>dt</strong></li>
</ul>
The three above can be combined as &rarr; <strong>dN/dt</strong>
<ul>
    <li>is dependent on &rarr; <strong>&equals;</strong></li>
    <li>the number of bacteria that actually divide - which we determined to be dependent on:</li>
    <ul>
        <li>the number of bacteria present at some time point &rarr; <strong>N(t)</strong></li>
        <li>the average rate of division &rarr; <strong>k</strong></li>
    </ul>
</ul>

The above statement can be rewritten into the below mathematical expression:

```
dN/dt = N(t)*k
```

Which gives us an equation that we can rewrite into:

```
N(t + dt) = N(t) + N(t)*k*dt
```

This equation now tells us that the number of bacteria after some time <strong>N(t + dt)</strong>, can be calculated by taking into account the number of bacteria at the initial timepoint <strong>N(t)</strong> and their rate of division <strong>k</strong> and how much time passed <strong>dt</strong>. 

And this set up seems again awfully similar to a for-loop. Let's compare it to a linear equation:

```
y(x) = m*x + b
```

Earlier in this notebook, you were able to translate your linear expression into something Python can interpret, and here, you can do the same with your equation for bacterial growth:

```python
N[t + dt] = N[t] + N[t]*k*dt
```

Something to consider is that we need to define some values beforehand, not too different than how you had to define your constants for your equation for a line and polynomial. Since we can't run a simulation forever, we'll need to decide how long we'll let our <i>in silico</i> bacteria to divide <code>t_max</code>, and we'll need to decide on how often we want to "monitor" our <i>in silico</i> bacteria <code>dt</code>. Then, we can consider the parameters of our bacteria, specifically the starting number of bacteria <code>N&lbrack;0&rbrack;</code> and their average rate of division <code>k</code>.

We can try with the following parameters to start with:

```python
t_max = 200 #minutes
dt = 1 #minute

N[0] = 1 #bacteria
k = 0.02 #divisions per minute
```

One additional thing to consider is given a simulation runtime of <code>t_max</code> and a frequency of <code>dt</code>, how does that then influence how you set up your for-loop?

`````python
#Let's set up a bacteria growth simulation for exponential growth
#First we set up our parameters
#We set up how long our simulation will run (t_max)
t_max = 200 #minutes
#Then we specify how often we will "observe" our bacteria
#Basically how many calculations our simulation will perform
dt = 1 #minute

#Then we need an array to hold the outputs
#So we initialize an array of zeros based on how many calculations we will perform
#which is determined by the simulation time and our step size
N = np.zeros(int(t_max/dt))

#And we can initialize how much bacteria we want to start with
N[0] = 1 #bacteria

#Then we specify how fast our bacteria will divide as a rate, so per units time
k = 0.02 #divisions per minute


#Then we can set up our for-loop to solve our equation/run our simulation
for t in range(0, t_max - dt, dt):
    N[t + dt] = N[t] + N[t]*k*dt
`````

Once you've run your model for bacterial growth, plot the result.

`````python
#Then we can plot the results of our simulation
plt.plot(N)

plt.xlim(0)
plt.ylim(0)
plt.xlabel('Time (min)')
plt.ylabel('Number of bacteria')

sns.despine()
`````

    <li><code>k</code></li>

As you play around with your parameters, see how the resulting simulation changes when you plot the results.

`````python
#Here we can play around with our different parameters
#namely our simulation time, how many bacteria we start with and how fast they divide
#And then we can see how the results of our simulation changes by changing these parameters
t_max = 100 #minutes
dt = 1 #minute

N = np.zeros(int(t_max/dt))
N[0] = 100 #bacteria

k = 0.1 #divisions per minute

for t in range(0, t_max - dt, dt):
    N[t + dt] = N[t] + N[t]*k*dt
`````
