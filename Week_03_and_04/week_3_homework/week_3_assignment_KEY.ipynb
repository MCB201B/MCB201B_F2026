<h2>Question 3</h2>

2pts

Return a stacked bar plot for any 2x2 contigency table returned by <code>question_1(df, gene)</code>. The Y-axis of the bar plot should be the percent of injection sites, where the total number of injection sites for each condition can be calculated by summing up the number of engrafted and not engrafted tumors.

Your stacked bar plot should also contain the following:

<ul>
    <li>Appropriate X-tick labels</li>
    <li>Appropriate Y-axis label</li>
    <li>Y-axis as percent values rather than a proportion or absolute counts</li>
    <li>Y-axis with 0 set to the origin</li>
    <li><code>'engraft'</code> bars colored as red or similar color</li>
    <li><code>'not_engraft'</code> bars colored as grey</li>
    <li>Appropriate annotations using the results of your statitiscal analysis from question 2</li>
</ul>

You do not need to regenerate a contingency table nor rerun a Fisher's exact test for this question. 

The evaluator will first run <code>question_1(df, gene)</code>, immediately pass the returned object from <code>question_1(df, gene)</code> to <code>question_2(contingency_table)</code>, and then pass both returned objects from question 1 and question 2 to evaluate your answer to this question before looping back to question 1 to continue with the evaluation.

```python
####################################################
## THIS CODE CELL WILL BE JUDGED BY THE EVALUATOR ##
####################################################

def question_3(contingency_table, pvalue):
    ### DO NOT MODIFY ANYTHING ABOVE THIS LINE ###
    ### YOUR ANSWER MUST BE BELOW THIS LINE WITHIN THIS CODE CELL ####

    fig, ax = plt.subplots()

    engraft_table_T = contingency_table.div(contingency_table.sum(axis=0), axis=1).T * 100
    
    x = engraft_table_T.index
    y = engraft_table_T.columns
    
    bottom=np.zeros(len(x))
    
    use_color=['#E64B35', '#777777']
    use_labels=['Engrafted', 'Not engrafted']
    
    for i in range(0, len(y), 1):
        plt.bar(x, 
                engraft_table_T[y[i]], 
                label=use_labels[i], 
                bottom=bottom, 
                color=use_color[i],
                lw=0.5, 
                edgecolor='k',
               )
        bottom += engraft_table_T[y[i]]

    x1, x2 = 0, 1
    y_max = 100
    adjustment = y_max*0.05
    h = adjustment * 0.5
    y = y_max + adjustment
    
    plt.plot([x1, x1, x2, x2], [y, y+h, y+h, y], lw=1, c='k')
    
    if pvalue < 0.0005: 
        pvalue_text = '***'
    elif pvalue < 0.005: 
        pvalue_text = '**'
    elif pvalue < 0.05: 
        pvalue_text = '*'
    else:
        pvalue_text = 'ns'
    
    plt.text((x1+x2)*0.5, y+(h*1.5), pvalue_text, ha='center', va='bottom', color='k', size=12)
    
    
    plt.ylabel('Engraftment (%)')
    plt.xticks([0, 1], 
               contingency_table.columns,
               rotation=35,
               ha='right',
              )
    
    plt.legend(loc='center', 
               bbox_to_anchor=(1.25,0.5), 
               fontsize=5, 
               edgecolor='white',
              )
    
    fig.set_dpi(300)
    fig.set_size_inches(2,3)
    
    sns.despine()
    plt.close()

    return fig
```
